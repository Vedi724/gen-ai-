```python
num = int(input('enter the value'))
sum_even = 0
sum_odd = 0
for i in range(1,num+1):
  if i % 2 == 0:
    sum_even += i
  else:
    sum_odd += i
print("even sum", sum_even)
print("odd sum ", sum_odd)
```
